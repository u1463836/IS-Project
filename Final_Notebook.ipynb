# 2. Business Problem & Context

## Description of the Organization
Blumhouse Productions is a film and TV production company based in Los Angeles. Jason Blum started it in 2000, and it's known for horror movies made on very small budgets, like Paranormal Activity, Get Out, The Purge, and M3GAN. Blumhouse keeps costs low by having actors work for basic union pay, then giving them a share of the profits if the movie does well. Directors get a lot of creative freedom, and the movies get wide releases through big studios like Universal. Blum explained the thinking: "when you have less budget, you have less risk and you have more fun" (Film Independent).

## Business Challenge
For years this model worked really well. But according to The Hollywood Reporter, all four Blumhouse movies released in the first half of 2025 (Wolf Man, The Woman in the Yard, Drop, and M3GAN 2.0) were major box office misses. M3GAN 2.0 opened to only $10.2 million, less than half of what the first M3GAN opened to. TheWrap called it an 18 month slump for the studio. Blum himself has admitted the box office can't hold as many horror movies as it used to. At the same time, Black Phone 2 and Five Nights at Freddy's 2 still did well later in the year. So, the challenge is that Blumhouse has to make big money decisions before a movie comes out. It has to choose how much to spend on marketing, how many theaters to open in, and what weekend to release it. But it doesn't find out if the movie is a hit or a flop until after all that money is already spent. Our goal is to give Blumhouse a way to predict that earlier.

## Business Impact
Even cheap movies cost more than their budget. The budget number usually only covers making the movie, not marketing, and theaters keep part of every ticket sold. For a small horror movie, marketing can cost as much as making it. As Variety put it, "Blumhouse avoids any total financial fiascos by keeping costs low" (Variety, 2025). But a flop still wastes marketing money and theater space, and a string of flops hurts the Blumhouse name with audiences. If Blumhouse could spot likely flops earlier, it could spend less marketing them, move their release dates, or send them to streaming. If it could spot likely hits, it could give them wider releases and bigger campaigns.

## Goals of the Analysis
We are using supervised learning to predict whether a movie will be a hit or a flop. Our target column is hit_flop. A movie counts as a hit (1) if its worldwide revenue is at least 2 times its budget, and a flop (0) if it isn't. We use 2 times because the budget doesn't include marketing, and theaters keep a big share of ticket sales. We only use info Blumhouse would have before release: budget, runtime, genre, release month, MPAA rating, number of theaters, Google search interest before release, and trailer views, likes, and comments. We trained the models on Blumhouse, Screen Gems, and A24 movies to get more data, but we interpret the results for Blumhouse.

## Relevant Industry Factors
Horror is doing really well right now. Sherwood News, using data from The Numbers, found that horror's share of US ticket sales hit an all time high of 17.2% in October 2025. The Associated Press reported that horror passed $1 billion at the US box office by early September 2025, which usually takes a full year. But more horror also means more competition, since Blumhouse is now fighting other studios' horror movies for the same weekends and the same audience. Reuters also reported that half of horror movies from major US distributors made at least 50% of their money outside the US, so international audiences matter too. A lot of movie marketing has also moved online, which is why we used trailer views and Google searches to measure buzz. We also wanted to use data from social media like TikTok, Instagram, and X, since that's where a lot of horror buzz happens now. But those sites either charge a lot for their data or don't let the public pull it, so we stuck with YouTube and Google Trends, which are free and easy to access.

# 3. Data Sources & References

## Brief Explanation of the Datasets
**TMDB API:** The Movie Database has info on almost every released movie. We pulled all movies made by Blumhouse, Screen Gems, and A24 using each studio's company ID. For each movie we got the title, budget, revenue, release date, runtime, genres, US rating, and audience rating and popularity scores. We only kept movies that had both a budget and revenue listed.

**Google Trends:** Google Trends shows how popular a search was on a scale of 0 to 100. 100 is the busiest point for that search, and the rest are compared to it. It doesn't give the actual number of searches. We pulled the average US search interest for each movie title in the 45 days before release and the 45 days after. Since each movie is scored against its own peak, this shows when buzz picked up for a movie, but it can't tell us which movie was searched more overall.

**YouTube Data API:** We searched YouTube for each movie's official trailer and pulled its views, likes, and comments. This shows how much people engaged with the marketing. YouTube only lets you do about 100 searches a day, so we had to pull this data over a few days.

**The Numbers:** The Numbers is a box office tracking site. We used it to double check budget and box office numbers from TMDB, since some movies had numbers that looked wrong.

**Box Office Mojo:** Box Office Mojo tracks box office earnings for every movie. We pulled each movie's domestic opening weekend, domestic and international box office, and the most theaters it played in. It doesn't have an API, so we read the info off each movie's page using web scraping.

## API Keys

```python
TMDB_API_KEY = userdata.get("TMDB_API_KEY")
YOUTUBE_API_KEY = userdata.get("YOUTUBE_API_KEY")
```

## TMDB: Movie list for each studio

```python
studios = {"Blumhouse": 3172, "Screen Gems": 3287, "A24": 41077}

movie_list = []
for studio, company_id in studios.items():
    page = 1
    total_pages = 1
    while page <= total_pages:
        url = "https://api.themoviedb.org/3/discover/movie"
        params = {"api_key": TMDB_API_KEY, "with_companies": company_id, "page": page}
        data = requests.get(url, params=params).json()
        total_pages = data["total_pages"]
        for m in data["results"]:
            movie_list.append({"studio": studio, "id": m["id"]})
        page += 1

print(len(movie_list), "movies found")
```

## TMDB: Details for each movie

```python
rows = []
for i, m in enumerate(movie_list):
    url = f"https://api.themoviedb.org/3/movie/{m['id']}"
    d = requests.get(url, params={"api_key": TMDB_API_KEY}).json()
    rows.append({
        "studio": m["studio"],
        "id": d["id"],
        "title": d["title"],
        "budget": d["budget"],
        "revenue": d["revenue"],
        "release_date": d["release_date"],
        "runtime": d["runtime"],
        "genres": [g["name"] for g in d["genres"]]
    })
    print(f"{i + 1}/{len(movie_list)} {m['studio']} - {d['title']}")
    time.sleep(0.1)

df = pd.DataFrame(rows)

# Keep only movies that have both a budget and revenue reported
df = df[(df["budget"] > 0) & (df["revenue"] > 0)].reset_index(drop=True)
print(df.shape)
df.head()
```

## Google Trends: Search interest 45 days before and after release

```python
pytrends = TrendReq(hl="en-US", tz=360)

def get_trend_avg(title, start, end):
    try:
        pytrends.build_payload([title], timeframe=f"{start} {end}", geo="US")
        data = pytrends.interest_over_time()
        return data[title].mean() if not data.empty else None
    except Exception:
        return None

pre, post = [], []
for i, row in df.iterrows():
    rd = datetime.strptime(row["release_date"], "%Y-%m-%d")
    before = (rd - timedelta(days=45)).strftime("%Y-%m-%d")
    after = (rd + timedelta(days=45)).strftime("%Y-%m-%d")
    pre.append(get_trend_avg(row["title"], before, row["release_date"]))
    post.append(get_trend_avg(row["title"], row["release_date"], after))
    print(f"{i + 1}/{len(df)} {row['title']}: pre={pre[-1]}, post={post[-1]}")
    time.sleep(2)

df["avg_trend_pre45"] = pre
df["avg_trend_post45"] = post
```

## YouTube: Trailer views, likes, and comments

```python
youtube = build("youtube", "v3", developerKey=YOUTUBE_API_KEY)

def get_trailer_stats(title):
    try:
        search = youtube.search().list(q=f"{title} official trailer", part="id", type="video", maxResults=1).execute()
        vid = search["items"][0]["id"]["videoId"]
        stats = youtube.videos().list(part="statistics", id=vid).execute()["items"][0]["statistics"]
        return int(stats.get("viewCount", 0)), int(stats.get("likeCount", 0)), int(stats.get("commentCount", 0))
    except Exception:
        return None, None, None

views, likes, comments = [], [], []
for i, row in df.iterrows():
    v, l, c = get_trailer_stats(row["title"])
    views.append(v)
    likes.append(l)
    comments.append(c)
    print(f"{i + 1}/{len(df)} {row['title']}: views={v}, likes={l}, comments={c}")
    time.sleep(1)

df["trailer_views"] = views
df["trailer_likes"] = likes
df["trailer_comments"] = comments
```

```python
youtube = build("youtube", "v3", developerKey=YOUTUBE_API_KEY)

def get_stats(video_ids):
    result = youtube.videos().list(part="statistics,snippet", id=",".join(video_ids)).execute()
    return result["items"]

def to_int(stats, key):
    return int(stats[key]) if key in stats else None

def find_trailer(title, year):
    searches = [f"{title} {year} official trailer", f"{title} movie trailer"]
    for q in searches:
        search = youtube.search().list(q=q, part="id", type="video", maxResults=5).execute()
        ids = [item["id"]["videoId"] for item in search["items"]]
        if not ids:
            continue
        videos = get_stats(ids)
        matches = [v for v in videos if title.lower().split(":")[0] in v["snippet"]["title"].lower()]
        if matches:
            best = max(matches, key=lambda v: int(v["statistics"].get("viewCount", 0)))
            s = best["statistics"]
            return to_int(s, "viewCount"), to_int(s, "likeCount"), to_int(s, "commentCount")
    return None, None, None

wrong = ["Five Nights at Freddy's", "The Wedding Ringer", "D.E.B.S.", "Swept Away", "Attack the Block", "Paranormal Activity 3"]
df.loc[df["title"].isin(wrong), "trailer_views"] = None

redo = df[df["trailer_views"].isna() | (df["trailer_views"] == 0)]
print(len(redo), "movies to run")

for i, row in redo.iterrows():
    try:
        v, l, c = find_trailer(row["title"], row["release_date"][:4])
    except Exception as e:
        if "quota" in str(e).lower():
            print("Hit the daily limit. Run again tomorrow.")
            break
        print("Error:", row["title"], e)
        continue
    df.loc[i, ["trailer_views", "trailer_likes", "trailer_comments"]] = [v, l, c]
    print(f"{i + 1}/{len(df)} {row['title']}: views={v}, likes={l}, comments={c}")
    time.sleep(1)
```

```python
# Add IMDb ID and audience rating columns from TMDB
imdb_ids, vote_avg, vote_count, popularity = [], [], [], []
for i, row in df.iterrows():
    d = requests.get(f"https://api.themoviedb.org/3/movie/{row['id']}", params={"api_key": TMDB_API_KEY}).json()
    imdb_ids.append(d.get("imdb_id"))
    vote_avg.append(d.get("vote_average"))
    vote_count.append(d.get("vote_count"))
    popularity.append(d.get("popularity"))
    print(f"{i + 1}/{len(df)} {row['title']}: {imdb_ids[-1]}")
    time.sleep(0.1)

df["imdb_id"] = imdb_ids
df["vote_average"] = vote_avg
df["vote_count"] = vote_count
df["popularity"] = popularity
```

```python
headers = {"User-Agent": "Mozilla/5.0"}

# Finds a label on the page (like "Domestic Opening") and returns the value next to it
def find_value(soup, label):
    for span in soup.find_all("span"):
        if span.text.strip().startswith(label):
            value = span.find_next_sibling("span")
            if value:
                return value.text.strip()
    return None

# Turns "$45,000,000" or "3,143 theaters" into a number
def to_number(text):
    if text is None:
        return None
    digits = "".join(c for c in text if c.isdigit())
    return int(digits) if digits else None

# US rating (R, PG-13) from TMDB
def get_rating(tmdb_id):
    url = f"https://api.themoviedb.org/3/movie/{tmdb_id}/release_dates"
    data = requests.get(url, params={"api_key": TMDB_API_KEY}).json()
    for country in data.get("results", []):
        if country["iso_3166_1"] == "US":
            for r in country["release_dates"]:
                if r["certification"]:
                    return r["certification"]
    return None

def get_mojo_data(imdb_id):
    try:
        # Main movie page: opening weekend and grosses
        page = requests.get(f"https://www.boxofficemojo.com/title/{imdb_id}/", headers=headers)
        soup = BeautifulSoup(page.text, "html.parser")
        data = {
            "opening_weekend": to_number(find_value(soup, "Domestic Opening")),
            "domestic_gross": to_number(find_value(soup, "Domestic (")),
            "international_gross": to_number(find_value(soup, "International (")),
            "widest_release": None
        }

        # Release page: widest release (number of theaters)
        link = soup.find("a", href=lambda h: h and h.startswith("/release/rl"))
        if link:
            release_url = "https://www.boxofficemojo.com" + link["href"].split("?")[0]
            release_page = requests.get(release_url, headers=headers)
            release_soup = BeautifulSoup(release_page.text, "html.parser")
            data["widest_release"] = to_number(find_value(release_soup, "Widest Release"))
        return data
    except Exception as e:
        print("Error:", e)
        return {}

# Run on all movies
mojo_rows = []
for i, row in df.iterrows():
    data = get_mojo_data(row["imdb_id"])
    data["mpaa_rating"] = get_rating(row["id"])
    mojo_rows.append(data)
    print(f"{i + 1}/{len(df)} {row['title']}: {data}")
    time.sleep(1)

# Add the new columns to df
df = pd.concat([df, pd.DataFrame(mojo_rows)], axis=1)
df.head()
```
